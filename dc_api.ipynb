```python
import dotenv
import json
dotenv.load_dotenv()

import os
import requests
import pandas as pd

RS_API_KEY = os.getenv("RUDDERSTACK_API_KEY")
RS_API_URL = "https://api.rudderstack.com"

headers = {
    "Authorization": f"Bearer {RS_API_KEY}",
    "Content-Type": "application/json",
}
```

---
## Events

```python
# create event
body = {
  "name": "SXH test event 11 ",
  "description": "Created with API",
  "eventType": "track",
  "categoryId": None
}

response = requests.post(f"{RS_API_URL}/v2/catalog/events", json=body, headers=headers )
print(response.json())
```

```python
# get events

response = requests.get(f"{RS_API_URL}/v2/catalog/events", headers=headers )

#print(response.json())
print(json.dumps(response.json()["data"], indent=4, sort_keys=True))
```

```python
# delete events
event_id = "ev_2xzHZ84lzRRb7Q2un3aH0BDEIjj"

response = requests.delete(f"{RS_API_URL}/v2/catalog/events/{event_id}", headers=headers )
#print(response.json())

if (response.status_code != 200):
    print(f"Error [{response.status_code}]: id:{event_id}  {response.json()['error']}")
```

## Properties

```python
# create property


body = {
  "name": "test-test-test2",
  "description": "where is this going",
  "type": "number"
}

response = requests.post(f"{RS_API_URL}/v2/catalog/properties", json=body, headers=headers )
print(response.json())
```

```python
# List properties

response = requests.get(f"{RS_API_URL}/v2/catalog/properties?page=1&orderBy=name:asc", headers=headers )
#print(json.dumps(response.json(), indent=4, sort_keys=True))
#print("*********************")

df_properties = pd.DataFrame(response.json()['data'])
if df_properties.empty:
    print("No properties found")
else:
    df_properties.index = df_properties['id']

df_properties
```

```python
response
```

```python
# Delete properites

if df_properties.shape[0] > 0:
    property_ids = df_properties['id'].tolist()
    print(property_ids)

    for property_id in property_ids:
        response = requests.delete(f"{RS_API_URL}/v2/catalog/properties/{property_id}", headers=headers)
        print(response.json())
else:
    print("No properties to delete")        



```
